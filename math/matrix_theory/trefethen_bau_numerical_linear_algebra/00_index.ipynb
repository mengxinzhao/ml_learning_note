# Numerical Linear Algebra — Detailed ML Systems Lecture Notes

## Trefethen & Bau, *Numerical Linear Algebra* (1997) — 40 Lectures

这套笔记严格沿着原书的 40 Lectures 走，但目标不是“摘要原书”，而是把每一讲整理成一份**可以独立读懂的数值线性代数讲义**，再把它映射到现代 ML systems。

每一讲尽量保持同一个结构：

1. **数学对象**：到底在算什么；
2. **关键公式**：最少保留能支撑工程判断的推导；
3. **几何/算法解释**：为什么这个算法是这样构造的；
4. **Numerical stability**：finite precision 下哪里会坏；
5. **ML systems mapping**：training / optimization / deployment 中对应什么；
6. **Python experiment**：用数值实验验证结论，而不是用代码代替数学。

证明不会像教材一样完整展开，但也不会跳过关键逻辑。

原书 PDF：https://www.stat.uchicago.edu/~lekheng/courses/309/books/Trefethen-Bau.pdf

## 阅读顺序

Part I–III 是主干，建议顺序读：Part I 建立 SVD / norm / 条件数的语言，Part II 给出 QR 与最小二乘，Part III 才把 conditioning 与 stability 这两件常被混为一谈的事分开。

Part IV–VI 相对独立，可按需取用：解线性方程组看 Part IV，特征值与 SVD 算法看 Part V，大规模迭代法（Arnoldi / GMRES / Lanczos / CG）看 Part VI。

## 贯穿全书的一个区分

不要把下面三件事混成一句“数值不稳”：

1. **operator amplification**：$\Vert A\Vert$ 大不大？
2. **problem conditioning**：$A^{-1}$ 是否敏感？
3. **algorithm stability**：实现是否额外放大 rounding error？

现代 ML numerics 中大量争论，其实是把这三个层次混在了一起。
